### Visualizations for Categorical Data

### Regression Model: Predicting CTC Offered

#### Handling Missing Values and Data Preparation for Regression

Before running the regression model, we need to address missing values, especially in the dependent variable ('CTC Offeredin lakhs PA') and the independent variables chosen for the model. We will create a new DataFrame by dropping rows that contain `NaN` values in these specific columns. This ensures that the regression analysis is performed on complete data, which is essential for accurate model fitting and interpretation.

```python
relevant_cols = [
    'CTC Offeredin lakhs PA',
    'Percentage_in_10_Class',
    'Percentage_in_12_Class',
    'Percentage_in_Under_Graduate',
    'Over all percentage in MBA(till now)'
]

df_cleaned_for_outliers = df.dropna(subset=relevant_cols).copy()

print(f"Original DataFrame shape: {df.shape}")
print(f"Cleaned DataFrame shape: {df_cleaned_for_outliers.shape}")

# Display info for the cleaned DataFrame to confirm no NaNs in relevant columns
df_cleaned_for_outliers[relevant_cols].info()
```

Now we can proceed with fitting the regression model using the `df_cleaned_for_outliers` DataFrame.

```python
import statsmodels.formula.api as smf

# Define the regression formula
# The dependent variable is 'CTC Offeredin lakhs PA'
# The independent variables are 'Percentage_in_10_Class', 'Percentage_in_12_Class', 'Percentage_in_Under_Graduate', and 'Over all percentage in MBA(till now)'
formula = 'Q("CTC Offeredin lakhs PA") ~ Q("Percentage_in_10_Class") + Q("Percentage_in_12_Class") + Q("Percentage_in_Under_Graduate") + Q("Over all percentage in MBA(till now)")'

# Create and fit the OLS (Ordinary Least Squares) model
# Using the df_cleaned_for_outliers DataFrame as it handles missing values in CTC and academic percentages
model = smf.ols(formula, data=df_cleaned_for_outliers)
results = model.fit()

# Print the model summary
print("Regression Model Summary:")
display(results.summary())
```

### Linearity Check: Residual Plots

```python
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm

# Get the fitted values and residuals from the OLS regression results
fitted_values = results.fittedvalues
residuals = results.resid

# --- Residuals vs. Fitted Values Plot ---
plt.figure(figsize=(10, 6))
sns.scatterplot(x=fitted_values, y=residuals)
plt.axhline(y=0, color='r', linestyle='--')
plt.title('Residuals vs. Fitted Values')
plt.xlabel('Fitted Values')
plt.ylabel('Residuals')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# --- Normal Q-Q Plot of Residuals ---
plt.figure(figsize=(8, 6))
sm.qqplot(residuals, line='s') # 's' for standardized line
plt.title('Normal Q-Q Plot of Residuals')
plt.show()
```

### Residual Plots against Independent Variables

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Independent variables (IVs) as used in the regression formula
iv_cols = [
    'Percentage_in_10_Class',
    'Percentage_in_12_Class',
    'Percentage_in_Under_Graduate',
    'Over all percentage in MBA(till now)'
]

# Ensure that the residuals are aligned with the original DataFrame's index
# The `df_cleaned_for_outliers` was used for regression, so align with its index
residuals_aligned = pd.Series(results.resid, index=df_cleaned_for_outliers.index)

plt.figure(figsize=(15, 10))

for i, col in enumerate(iv_cols):
    plt.subplot(2, 2, i + 1) # Arrange plots in a 2x2 grid
    sns.scatterplot(x=df_cleaned_for_outliers[col], y=residuals_aligned)
    plt.axhline(y=0, color='r', linestyle='--')
    plt.title(f'Residuals vs. {col}')
    plt.xlabel(col)
    plt.ylabel('Residuals')
    plt.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
